```python
# now we will make function to do all these things
def preprocess_text(text):
    # Convert text to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Tokenize the text
    tokens = word_tokenize(text)

    # Remove stopwords
    filtered_tokens = []

    for word in tokens:
        if word not in stop_words:
            filtered_tokens.append(word)

    # Join the words back into a single string
    cleaned_text = " ".join(filtered_tokens)

    return cleaned_text
```

```python
# lets test our function
test_text = "donald trump is speaking about his country"

clean = preprocess_text(test_text)

print("Original text:", test_text)
print("Clean text:", clean)
```

```python
# Adding extra cell in data set 
df["content"] = df["title"] + " " + df["text"]

print(df[["title" , "text" , "content"]].head())
```

```python
# test first 5 article

df["content"].head().apply(preprocess_text)
```

```python
# now we will save all process data in our data set in to clean_content column
df["clean_content"] = df["content"].apply(preprocess_text)

print(df[["content", "clean_content"]].head())
```

```python
# to check if there is any zero or missing in our data set
print(df["clean_content"].isnull().sum())
print((df["clean_content"] == "").sum())
```

```python
X = df["clean_content"]
y = df["label"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#NOW WE WILL IMPORT TRAIN TEST SPLIT FUNCTION FROM SK LEARN
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20, 
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# now we will import tfidf function from sklearn to convert text into numerical values
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=50000)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
# now time for logisticregression to train our model 
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000) 

model.fit(X_train_tfidf, y_train)

print("Model training completed!")
```

```python
# now we will se our model predictions
y_pred = model.predict(X_test_tfidf)

print("Predictions completed!")
print("First 10 predictions:")
print(y_pred[:10])
```

```python

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
 
print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
# now lets calculate precision accuracy and f1 score
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_test, y_pred, pos_label="Fake")
recall = recall_score(y_test, y_pred, pos_label="Fake")
f1 = f1_score(y_test, y_pred, pos_label="Fake")

print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
```

```python
#  time to confustion matricx to check our model acual prediction in tables
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred, labels=["Fake", "True"])

# to se how many true news predict fake by our model vise virsa
print("Confusion Matrix:")
print(cm)
```

```python
# a small custome test to see prediction 
news_text = input("Enter news article: ")

clean_news = preprocess_text(news_text)

news_tfidf = tfidf.transform([clean_news])

prediction = model.predict(news_tfidf)[0]

print("Prediction:", prediction)
```
